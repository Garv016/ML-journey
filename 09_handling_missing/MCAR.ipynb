## Converting cat to Num


1. Male Onehot 
2. relevent_exp Ordinal
3. enrolled_uni Ordinal
4. eduLevel Ordinal
5. major_discipline OneHot
6. company type One Hot

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

preprocessor = ColumnTransformer(
    transformers=[
        # One-Hot
        ('gender', OneHotEncoder(drop='first', handle_unknown='ignore'),
         ['gender']),

        ('major', OneHotEncoder(drop='first', handle_unknown='ignore'),
         ['major_discipline']),

        ('company_type', OneHotEncoder(drop='first', handle_unknown='ignore'),
         ['company_type']),

        # Ordinal
        ('relevent_exp',
         OrdinalEncoder(categories=[
             ['No relevent experience',
              'Has relevent experience']
         ]),
         ['relevent_experience']),

        ('enrolled_uni',
         OrdinalEncoder(categories=[
             ['no_enrollment',
              'Part time course',
              'Full time course']
         ]),
         ['enrolled_university']),

        ('education',
         OrdinalEncoder(categories=[
             ['Primary School',
              'High School',
              'Graduate',
              'Masters',
              'Phd']
         ]),
         ['education_level'])
    ],
    remainder='passthrough'
)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop('target',axis=1),df['target'],test_size=0.2)
X2_train, X2_test, y2_train, y2_test= train_test_split(df2.drop('target',axis=1),df2['target'],test_size=0.2)
```

```python
print(type(X_train))
print(type(X_test))
```

```python
X_train.head()
```

```python
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
X2_train = preprocessor.fit_transform(X2_train)
X2_test = preprocessor.transform(X2_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)

lr2 = LogisticRegression()
lr2.fit(X2_train,y2_train)
y2_pred = lr2.predict(X2_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("With NAN modified with mean and mode",accuracy_score(y_test,y_pred))
print("With NaN drop",accuracy_score(y2_test,y2_pred))
```
