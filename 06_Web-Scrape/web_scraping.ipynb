## Getting all Company, title , difficulty and duration together

```python
cards = webPretty.find_all(
    "li",
    class_="cds-grid-item"
)
titles = []
companies = []
levels = []
durations = []

for card in cards:

    # title
    title = card.find(
        "h3",
        class_="cds-CommonCard-title"
    )

    # company
    company = card.find(
        "p",
        class_="cds-ProductCard-partnerNames"
    )

    # level + duration
    meta = card.find(
        "div",
        class_="cds-CommonCard-metadata"
    )

    print("Title:", title.get_text(strip=True) if title else "")
    titles.append(title.get_text(strip=True))

    print("Company:", company.get_text(strip=True) if company else "")
    companies.append( company.get_text(strip=True))

    if meta:
        text = meta.get_text(" ", strip=True)

        parts = [p.strip() for p in text.split("Â·")]

        level = parts[1]
        levels.append(parts[1])
        duration = parts[-1]
        durations.append(parts[-1])

        print("Level:", level)
        print("Duration:", duration)

    # print("-" * 40)
```

```python
d = {"titles" : titles, "companies" : companies, "levels" : levels, "durations" : duration}
df = pd.DataFrame(d)
df
```
