# Fetching from TMDB website


Title, overview, release date, vote_averae,popularity

```python
import pandas as pd
import requests

all_movies = []

for page in range(1, 6):   # first 5 pages = 100 movies
    url = f"https://api.themoviedb.org/3/trending/movie/day?api_key=5b4fa3f8fdae4392396b6d470f42d6cd&page={page}"
    data = requests.get(url).json()

    all_movies.extend(data["results"])

df = pd.DataFrame(all_movies)[
    ['title','overview','release_date','vote_average','popularity']
]
all_movies
```
